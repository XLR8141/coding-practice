```python
import pandas as pd
import seaborn as sns
data = pd.read_csv(r"C:\Users\pc\Downloads\tested.csv")
data.head()
```

### 1. Scatter plot (Numerical - Numerical)

```python
sns.scatterplot( data = data, x = "Age",y = "Fare",style = "Embarked",hue = "Sex")
```

### 2. Bar chart (Numerical - catagorical)

```python
sns.barplot(data = data, x = "Embarked", y = "Fare")
```

### 3. Box plot (Numerical - catagorical)

```python
sns.boxplot(data = data, x = "Age",y = "Sex")
```

### 4. Dist plot (Numerical - catagorical)

```python
sns.distplot(data[data["Survived"] == 1]["Age"],hist = False)
sns.distplot(data[data["Survived"] == 0]["Age"],hist = False)
```

### Heatmap (catagorical - catagorical)

```python
sns.heatmap(pd.crosstab(data['Pclass'],data["Survived"]))
```

### Clustermap (catagorical - catagorical)

```python
sns.clustermap(pd.crosstab(data["Survived"],data["Pclass"]))

```
