```python
import pandas as pd 
import numpy as np
df = pd.read_csv("C:/Users/pc/Downloads/dirty_cafe_sales.csv")
df.head()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(" ","_")
df
```

```python
df["total_spent"]= df['total_spent'].replace("ERROR",None)
df
```

```python
df['total_spent'] = pd.to_numeric(df['total_spent'], errors='coerce')
df['quantity'] = pd.to_numeric(df['quantity'], errors='coerce')
df['price_per_unit'] = pd.to_numeric(df['price_per_unit'], errors='coerce')
```

```python
df['total_spent'] = df['total_spent'].fillna(df['quantity'] * df['price_per_unit'])
df
```

```python
df = df.drop(columns = "location")
df
```

```python
df["transaction_date"] = df["transaction_date"].replace("ERROR",0)
df["transaction_date"] = pd.to_datetime(df["transaction_date"], errors='coerce')
```

```python
df = df.dropna()
df.head(30)
```
