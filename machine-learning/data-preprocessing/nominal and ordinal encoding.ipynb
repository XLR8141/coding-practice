## Ordianl and Label encoding

. use ordinal encoding when data follows an order or ranking like bad < average < good   , can be used for both input and output features

. use label encoding only for only y feature lable encoding is used for nominal data which dosent follow any order or ranking



```python
import pandas as pd 
data = pd.read_csv(r"C:\Users\pc\Downloads\customer.csv")
data.head()
```

```python
from sklearn.preprocessing import OrdinalEncoder
od = OrdinalEncoder(categories=[["Poor","Average","Good"],["School","UG","PG"]])
data_scaled = od.fit_transform(data[["review","education"]])
data[["review","education"]] = data_scaled
```

```python
data.head()
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
data["purchased"] = le.fit_transform(data["purchased"])
data.head()
```

## a better way to apply all these encoding is to use Column transformer which is an efficient and faster way 
### . method    ColumnTransformer( Transformer = ( ( name , tranformer you want to use , [column name] ), remainder = 'passthrough' ))

```python
import pandas as pd
dataf = pd.read_csv(r"C:\Users\pc\Downloads\covid_toy.csv")
dataf
```

```python
dataf["cough"].value_counts()
```

```python
# import nessecery libraries 
from sklearn.preprocessing import OneHotEncoder,OrdinalEncoder
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer

tranform = ColumnTransformer([("tf1",OrdinalEncoder(categories= [["Mild","Strong"]]),["cough"]),
                   ("tf2",OneHotEncoder(sparse_output = False),["city"]),
                   ("tf3",SimpleImputer(strategy= "most_frequent"),["fever"])],
                   remainder= 'passthrough')
```

```python
tranformed_data = pd.DataFrame(tranform.fit_transform(dataf))
tranformed_data
```
