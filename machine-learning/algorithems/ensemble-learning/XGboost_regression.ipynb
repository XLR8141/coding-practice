| Parameter | Purpose | Typical Values |
| --- | --- | --- |
| ``n_estimators`` | Number of trees | 100–500 |
| ``max_depth`` | Maximum depth of each tree | 3–10 |
| ``learning_rate`` | Shrinkage factor (step size) | 0.01–0.3 |
| ``objective`` | Defines task type | ``"reg:squarederror"`` (regression), ``"binary:logistic"`` (binary classification), ``"multi:softmax"`` (multi-class) |
| ``num_class`` | Needed only for multi-class classification | Number of classes |

```python
import pandas as pd
from sklearn.datasets import fetch_california_housing

df = fetch_california_housing(as_frame=True)
data = df.frame
data                   
```

```python
from xgboost import XGBRegressor

model = XGBRegressor(n_estimators = 200,
                        max_depth=6,
                        learning_rate=0.1,
                        objective="reg:squarederror")
```

```python
from sklearn.model_selection import train_test_split
X = data.iloc[:,:-1]
y = data.iloc[:,-1]
X_train,X_test,y_train,y_test = train_test_split(X,y,random_state=42,test_size=0.2)
```

```python
model.fit(X_train,y_train)
```

```python
pred = model.predict(X_test)
```

```python
from sklearn.metrics import r2_score
r2_score(y_test,pred)
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np

# Predict
y_pred = model.predict(X_test)

# Evaluate
print("R²:", r2_score(y_test, y_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred)))
print("MAE:", mean_absolute_error(y_test, y_pred))
```

```python
dataset = pd.read_csv(r"C:\Users\pc\Downloads\covid_toy.csv",)
dataset
```

```python
# Convert all object columns in dataset to category
for col in dataset.select_dtypes(include="object").columns:
    dataset[col] = dataset[col].astype("category")

print(dataset.dtypes)  # confirm no 'object' left

dataset['has_covid'].replace({'Yes':1,'No':0},inplace=True)
```

```python
X = dataset.iloc[:,:-1]
y = dataset.iloc[:,-1]
X_train,X_test,y_train,y_test = train_test_split(X,y,random_state=42,stratify=y,test_size=0.2)
```

```python
from xgboost import XGBClassifier
model = XGBClassifier(n_estimators = 600,max_dept = 6,tree_method="hist",learning_rate = 0.3,objective = 'binary:logistic', enable_categorical=True)
```

```python
model.fit(X_train,y_train)
```

```python
model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
print("Accuracy:", accuracy_score(y_test, model.predict(X_test)))
from sklearn.metrics import confusion_matrix
print("Confusion Matrix:\n", confusion_matrix(y_test, model.predict(X_test)))
from sklearn.metrics import classification_report
print(classification_report(y_test, model.predict(X_test)))

```
