```python
from sklearn.datasets import load_iris
import pandas as pd 
data = pd.read_csv(r"C:\Users\pc\Downloads\covid_toy.csv")
data['cough'].value_counts()
print(data)
```

```python
data.drop(columns=['city'],inplace=True)
data['cough'].replace({'Mild':0,'Strong':1},inplace=True)
data['gender'].replace({'Male':1,'Female':0},inplace=True)
data['has_covid'].replace({'No':0,'Yes':1},inplace=True)
data
```

```python
from sklearn.model_selection import train_test_split
X = data.iloc[:,:3]
y = data.iloc[:,-1]
X_train,X_test,y_train,y_test = train_test_split(X,y,random_state= 2,stratify=y)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import r2_score

lr = LogisticRegression()
lr.fit(X_train,y_train)
lr_pred = lr.predict(X_test)

dt = DecisionTreeClassifier(max_depth=5)
dt.fit(X_train,y_train)
dt_pred = dt.predict(X_test)

knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train,y_train)
knn_pred = knn.predict(X_test)


print ("logistic regesion :",r2_score(y_test,lr_pred))
print ("Decision tree :",r2_score(y_test,dt_pred))
print ("Knearest neighbour :",r2_score(y_test,knn_pred))
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import VotingClassifier

lr = LogisticRegression()
dt = DecisionTreeClassifier(max_depth= 5)
knn = KNeighborsClassifier(n_neighbors=5)

ensemble = VotingClassifier(estimators=[('logistic',lr),('decisiontree',dt),('kneighbour',knn)],voting = 'soft')
ensemble.fit(X_train,y_train)
ensemble_pred = ensemble.predict(X_test)
```

```python
from sklearn.metrics import r2_score
r2_score(y_test,ensemble_pred)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Accuracy
print("Accuracy:", accuracy_score(y_test, ensemble_pred))

# Detailed classification report
print("\nClassification Report:\n", classification_report(y_test, ensemble_pred))

# Confusion matrix
print("\nConfusion Matrix:\n", confusion_matrix(y_test, ensemble_pred))
```
