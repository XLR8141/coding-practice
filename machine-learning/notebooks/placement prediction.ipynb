```python
import pandas as pd 
data = pd.read_csv(r"C:\Users\pc\Downloads\placement.csv")
data
```

```python
data.info()
```

```python
#cleaning the dataset
data = data.iloc[:,1:]
data
```

```python
#seeing the graph of the dataset using scatter plot

import matplotlib.pyplot as plt

color_map = { 1:'green' , 0:'red'}
colors = data['placement'].map(color_map)

plt.scatter(data['cgpa'],data['iq'], c=colors)
plt.xlabel('cgpq')
plt.ylabel('iq')
plt.title('placement graph')
```

```python
# selecting data and the label for the model

X = data.iloc[:,0:2]
y = data.iloc[:,-1]
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.2)
```

```python
from sklearn.preprocessing import StandardScaler
scale = StandardScaler()
X_train_scaled = scale.fit_transform(X_train)
X_test_scaled = scale.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(X_train_scaled,y_train)
```

```python
predict = model.predict(X_test_scaled)
```

```python
y_test
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,predict)
```
