```python
import pandas as pd
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error
from sklearn.model_selection import train_test_split
data_set = pd.read_csv(r"C:\Users\pc\Downloads\melb_data.csv")

data_set
```

```python
data_set.columns
```

```python
data_set = data_set.dropna()
y = data_set.Price
features = ['Landsize', 'YearBuilt', 'BuildingArea', 'Rooms', 'Bathroom', 'Bedroom2']
feature = data_set[features]
X = feature
```

```python
train_X, val_X, train_y, val_y = train_test_split(X, y, random_state=1)
alg = DecisionTreeRegressor(random_state=1)
alg.fit(train_X, train_y)
```

```python
print(f"Predictions for first 5 rows: {alg.predict(X.head())}")
print(f"Actual prices: {y.head().tolist()}")
```

```python
prediction = alg.predict(val_X)
mae = mean_absolute_error(val_y, prediction)
print("Mean Absolute Error:", mae)
```

```python
def mae_(max_leaf_node,train_X,val_X,train_y,val_y):
    model = DecisionTreeRegressor(max_leaf_nodes= max_leaf_node,random_state = 1)
    model.fit(train_X,train_y)
    prediction = model.predict(val_X)
    mae = mean_absolute_error(val_y,prediction)
    return mae
```

```python
for max_leaf_nodes in [5,50,500,5000,50000]:
    my_mae = mae_(max_leaf_nodes,train_X,val_X,train_y,val_y)
    print(f"Max leaf nodes: {max_leaf_nodes}\t\tMean Absolute Error: {my_mae}")
```
