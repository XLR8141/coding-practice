```python
# webscrapping

import pandas as pd 
import requests
from bs4 import BeautifulSoup

web_page = requests.get('https://books.toscrape.com/')
soup = BeautifulSoup(web_page.text,'lxml')
```

```python
# Title

soup.find_all('h3')
```

```python
for i in soup.find_all('h3'):
    print(i.text)
```

```python
# for stocks
soup.find_all('p',class_='instock')
```

```python
# price

price = soup.find_all('p',class_="price_color")
for p in price:
    print(p.text.strip('Â'))
```

```python
title = []
price = []
rating = []
stock = []

col = soup.find_all("article", class_="product_pod")

for index in col:
    title.append(index.find('h3').find("a")["title"])
    price.append(index.find('p',class_ = 'price_color').text.strip('Â'))
    rating.append(index.find('p',class_ = 'star-rating')["class"][1])
    stock.append(index.find('p',class_ = 'instock').text.strip())
```

```python
print("Titles:", title)
print("Prices:", price)
print("Ratings:", rating)
print("stock:", stock)
```

```python
# creating dataframe
import pandas as pd

data = {"Title":title,"Prices":price,"Ratings":rating,"stock":stock}
df = pd.DataFrame(data)
df
```

```python
title = []
price = []
rating = []
stock = []

for page in range(1,51):
    web_page = requests.get(f'https://books.toscrape.com/catalogue/page-{page}.html')
    soup = BeautifulSoup(web_page.text,'lxml')
    pages = soup.find("article", class_="product_pod")
    
    col = soup.find_all("article", class_="product_pod")

    for index in col:
        title.append(index.find('h3').find("a")["title"])
        price.append(index.find('p',class_ = 'price_color').text.strip('Â'))
        rating.append(index.find('p',class_ = 'star-rating')["class"][1])
        stock.append(index.find('p',class_ = 'instock').text.strip())

```

```python
print("Titles:", title)
print("Prices:", price)
print("Ratings:", rating)
print("stock:", stock)
```

```python
products = {"Title":title,"Prices":price,"Ratings":rating,"stock":stock}
products_df = pd.DataFrame(products)
products_df
```
